```python
# ---------------------------------------------
# ETAPE 1 : ANALYSE EXPLORATOIRE DES BATIMENTS
# ---------------------------------------------
print("Étape 1 : Début de l'analyse exploratoire\n")

# Imports principaux pour l'analyse exploratoire
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# %matplotlib inline  # Permet d'afficher les graphiques directement dans le notebook au lieu d'une fenêtre separer.
%matplotlib inline
```

```python
# 1. Chargement du fichier CSV
# On indique le chemin du fichier. Ici, il est dans le dossier 'data' à la racine du projet.
fichier_csv = "../data/2016_Building_Energy_Benchmarking.csv"

# On lit le CSV avec pandas pour obtenir un DataFrame, qui est une structure de données similaire à une table Excel
batiments = pd.read_csv(fichier_csv)
print(f"CSV chargé depuis : {fichier_csv}")
print(f"Nombre de bâtiments total : {batiments.shape[0]}")
print("Colonnes disponibles :", batiments.columns.tolist())
print("Affichage du csv 3 premières lignes :")

display(batiments.head(3))  # Affiche les 3 premières lignes pour vérifier

# 2. Aperçu initial des données
# .head() affiche les 5 premières lignes du tableau pour avoir une idée des colonnes et valeurs
# .info() permet de vérifier le nombre de lignes, les types de colonnes et s'il y a des valeurs manquantes
print("\nInformations générales sur le jeu de données :")
batiments.info()
```

```python
# 3. Sélection des colonnes pertinentes
# On ne conserve que les colonnes utiles pour l'analyse et la modélisation
colonnes_pertinentes = [
    'PropertyName',             
    'PrimaryPropertyType',     
    'Address',                  
    'YearBuilt',                
    'PropertyGFABuilding(s)',  
    'SiteEUI(kBtu/sf)',        
    'SiteEnergyUse(kBtu)',     
    'TotalGHGEmissions',    
    'Latitude',            
    'Longitude',
    'Neighborhood',
    'ListOfAllPropertyUseTypes',
    'Electricity(kBtu)',
    'NaturalGas(kBtu)',
    'SteamUse(kBtu)'
]
batiments = batiments[colonnes_pertinentes]
print("\nColonnes conservées :", batiments.columns.tolist())
display(batiments.head(3))

```

```python
# 4. Définition des types utiles basés sur la fréquence observée dans les données
# Ici, nous sélectionons les type les plus utiles pour la modalisation 
# fiable dans la colonne PrimaryPropertyType  
types_utiles = [
    'Low-Rise Multifamily',        # Très fréquent : données nombreuses et homogènes
    'Mid-Rise Multifamily',        # Très fréquent : cohérent pour l'analyse
    'High-Rise Multifamily',       # Suffisamment fréquent et homogène
    'Small- and Mid-Sized Office', # Catégorie bien représentée et cohérente
    'Large Office',                # Bureaux de grande taille, assez fréquents
    'Warehouse',                   # Nombre suffisant pour analyse (entrepôts)
    'Retail Store',                # Magasins, assez nombreux pour être utilisables
    'Hotel',                       # Catégorie homogène avec nombre acceptable
    'K-12 School'                  # Écoles, assez de données et cohérentes
]

print("Types utiles sélectionnés pour le filtrage :")
print(types_utiles)
print("\n---------------------------------------------\n")

# Nombre de bâtiments avant filtrage
# .shape[0] donne le nombre total de lignes (bâtiments) dans le DataFrame
avant = batiments.shape[0]
print(f"Nombre de bâtiments AVANT filtrage : {avant}")

# Filtrage des bâtiments selon les types utiles
# isin() vérifie si la valeur de PrimaryPropertyType appartient à la liste types_utiles
# Le DataFrame ne garde que les bâtiments dont le type est dans la liste
batiments = batiments[batiments['PrimaryPropertyType'].isin(types_utiles)]

# Nombre de bâtiments après filtrage
apres = batiments.shape[0]
print(f"Nombre de bâtiments APRÈS filtrage : {apres}")
print(f"Bâtiments conservés : {apres} sur {avant}")
```

```python
# 5. Filtrage des bâtiments aberrants
# On supprime les bâtiments avec des valeurs impossibles ou incohérentes
avant_filtrage = batiments.shape[0]
batiments = batiments[(batiments['PropertyGFABuilding(s)'] > 0) &
                      (batiments['SiteEUI(kBtu/sf)'] > 0) &
                      (batiments['SiteEnergyUse(kBtu)'] > 0)]
apres_filtrage = batiments.shape[0]
print(f"\nNombre de bâtiments après filtrage : {apres_filtrage} (avant : {avant_filtrage})")
display(batiments.head(3))
```

```python
# 6. Vérification des valeurs manquantes
# On compte combien de valeurs sont manquantes dans chaque colonne
#Detection des valeurs manquantes avec .isna() il crée un tableau de True/False qui dit si chaque valeur est :
#True  manque
#False  non manquante
#Cela permet d'avoir un visuel sur les colonnes avec les valeur manquants et donc savoir quelles colonnes serais utilisables
print("\nNombre de valeurs manquantes par colonne :")
display(batiments.isna().sum())
```

```python
# 7. Choix de la target
# Pour la suite du projet, il faut qu'on choisisse une colonne à prédire.
# Ici on choisit 'SiteEnergyUse(kBtu)', car elle représente la consommation totale d’énergie du bâtiment
target = 'SiteEnergyUse(kBtu)'
print(f"\nTarget choisie pour la modélisation : {target}")

# 8. Statistiques descriptives
# On regarde la moyenne, le min, max, quartiles, etc., de la target
print("\nStatistiques descriptives de la target :")
display(batiments[target].describe())
```

```python
# 9. Visualisation : distribution de la target
# On utilise un histogramme pour voir comment se répartissent les consommations
plt.figure(figsize=(8,5))
sns.histplot(batiments[target], bins=30, kde=True)  # kde= en True ajoute une courbe de densité
plt.xlabel('Consommation d\'énergie (kBtu/ft²)')
plt.ylabel('Nombre de bâtiments')
plt.title('Distribution de la consommation d\'énergie')
plt.show()

# 10. Visualisation : consommation selon le type de bâtiment
# Boxplot : pratique pour voir les valeurs extrêmes et comparer plusieurs catégories
plt.figure(figsize=(12,6))
sns.boxplot(x='PrimaryPropertyType', y=target, data=batiments)
plt.xticks(rotation=45, ha='right')
plt.title('Consommation d\'énergie par type de bâtiment')
plt.show()
```

```python
from sklearn.preprocessing import OneHotEncoder

# ---------------------------------------------
# ETAPE 2 : FEATURE ENGINEERING
# ---------------------------------------------

print("\nÉtape 2 : Feature Engineering\n")

# 1. Création de la feature BuildingAge :
# On calcule l'âge du bâtiment en soustrayant son année de construction (YearBuilt)
# à l'année de référence (ici 2016).
# Le but est de transformer YearBuilt (difficile à interpréter) en une information plus utile :
# l'âge réel du bâtiment.
# Cette feature aide le modèle car : 
#  - un bâtiment vieux consomme souvent plus d'énergie,
#  - les équipements anciens sont moins efficaces,
#  - cela crée une variable numérique facile à analyser.
batiments['BuildingAge'] = 2016 - batiments['YearBuilt']
print("Feature 'BuildingAge' ajoutée : âge du bâtiment calculé à partir de YearBuilt")
display(batiments[['YearBuilt','BuildingAge']].head(3))
```

```python
# 2. Création de la feature BuildingSize :
# On copie la colonne PropertyGFABuilding(s) dans une nouvelle colonne au nom plus simple.
# Cette duplication permet :
#  - de garder la colonne originale intacte,
#  - d'utiliser BuildingSize pour des transformations (scaling, normalisation),
#  - de créer plus facilement des ratios basés sur la taille du bâtiment,
#  - d'avoir un nom plus clair et cohérent pour l'entraînement du modèle.
batiments['BuildingSize'] = batiments['PropertyGFABuilding(s)']  # Surface totale du/des bâtiments
print("Feature 'BuildingSize' ajoutée : même valeur que PropertyGFABuilding(s) pour clarté et futur usage")
display(batiments[['PropertyGFABuilding(s)','BuildingSize']].head(3))
```

```python
# 3. One-hot encoding pour les types de bâtiments avec OneHotEncoder :
# On transforme la colonne PrimaryPropertyType (texte) en plusieurs colonnes 0/1.
# Exemple : PrimaryPropertyType = "Hotel" → colonne PrimaryPropertyType_Hotel = 1, les autres colonnes = 0.
# Pourquoi faire ça ?
#  - Les modèles de machine learning ne comprennent pas le texte.
#  - Chaque type devient une information numérique exploitable par le modèle.
#  - Cela permet au modèle de capturer les différences entre types de bâtiments
#    (un hôtel ne consomme pas comme une école ou un bureau).
types_utiles = [
    'Hotel',
    'K-12 School',
    'Small- and Mid-Sized Office',
    'Large Office',
    'Mixed Use Property',
    'Medical Office',
    'Senior Care Community',
    'Retail Store',
    'University'
]

# On ne garde que les lignes correspondant aux types utiles
batiments_subset = batiments[batiments['PrimaryPropertyType'].isin(types_utiles)]

# Initialisation de l'encodeur OneHotEncoder de scikit-learn
# sparse=False : on veut un array dense (pandas préfère ça pour concaténer)
# drop=None : on ne supprime aucune catégorie, on garde toutes les catégories listées
encoder = OneHotEncoder(sparse=False, drop=None)

# On applique l'encodeur sur la colonne PrimaryPropertyType
# fit_transform fait deux choses en même temps :
# 1) 'fit' : l'encodeur apprend quelles sont toutes les catégories présentes
# 2) 'transform' : il transforme chaque catégorie en colonnes binaires 0/1
encoded = encoder.fit_transform(batiments_subset[['PrimaryPropertyType']])

# Récupération des noms des colonnes créées automatiquement
# Chaque nom correspond à une catégorie encodée, ex : PrimaryPropertyType_Hotel
encoded_columns = encoder.get_feature_names_out(['PrimaryPropertyType'])

# Conversion de l'array numpy encodé en DataFrame pandas
# On utilise les noms de colonnes générés et les mêmes indices que le DataFrame original
encoded_df = pd.DataFrame(encoded, columns=encoded_columns, index=batiments_subset.index)

# Supprimer d'abord les colonnes déjà créées pour éviter les doublons
for col in encoded_df.columns:
    if col in batiments.columns:
        batiments.drop(col, axis=1, inplace=True)

# Ajout des nouvelles colonnes binaires au DataFrame original
# Chaque colonne correspond à un type de bâtiment et contient 0 ou 1
batiments = pd.concat([batiments, encoded_df], axis=1)

print(f"Colonnes encodées ajoutées : {list(encoded_columns)}")
display(batiments[['PrimaryPropertyType'] + list(encoded_columns)].head(5))
```

```python
# 4. Features géographiques :
# On conserve les colonnes Latitude et Longitude pour chaque bâtiment.
# Pourquoi on les conserves car ?
#  - Les modèles de machine learning ne comprennent pas le texte, ils ont besoin de nombres.
#  - Transformer en float permet d’utiliser ces valeurs dans des calculs numériques.
#  - Ces colonnes permettent au modèle de capturer des différences liées à la localisation
#    (par exemple, des bâtiments dans des zones froides ou chaudes peuvent consommer différemment).
batiments['Latitude'] = batiments['Latitude'].astype(float)
batiments['Longitude'] = batiments['Longitude'].astype(float)
print("Features 'Latitude' et 'Longitude' conservées pour localisation")
```

```python
# 5. Features sur usage multiple
# On crée :
# NbPropertyUses → nombre total d'usages pour chaque bâtiment
# Has_<Usage> → colonnes binaires pour certains usages importants
# Exemple :
#  ListOfAllPropertyUseTypes = "Office, Retail, Parking" 
#  → NbPropertyUses = 3
#  → Has_Office = 1, Has_Retail = 1, Has_Parking = 1, Has_Restaurant = 0
# Pourquoi faire ça ?
#  - Les modèles de machine learning ne comprennent pas les listes/textes directement.
#  - NbPropertyUses capture l'effet du multi-usage global.
#  - Les colonnes Has_<Usage> capturent les usages spécifiques importants, 
#    permettant au modèle de détecter des différences entre types d'usages.

#  Nettoyer les espaces et mettre tout en majuscules pour uniformiser
batiments['ListOfAllPropertyUseTypes'] = batiments['ListOfAllPropertyUseTypes']\
    .fillna('')\
    .apply(lambda x: ','.join([u.strip() for u in x.split(',')]))

#  Créer NbPropertyUses
batiments['NbPropertyUses'] = batiments['ListOfAllPropertyUseTypes']\
    .apply(lambda x: len(x.split(',')) if x != '' else 1)

#  Définir les usages importants
usages_importants = ['Multifamily Housing', 'Parking', 'Office', 'Retail Store',
                     'Non-Refrigerated Warehouse', 'K-12 School', 'Restaurant', 
                     'Hotel', 'Swimming Pool']

#  Créer les colonnes binaires correctement
for usage in usages_importants:
    batiments[f'Has_{usage.replace(" ","_")}'] = batiments['ListOfAllPropertyUseTypes']\
        .apply(lambda x: 1 if usage in x.split(',') else 0)

#  Vérification
display(batiments[['ListOfAllPropertyUseTypes','NbPropertyUses'] + 
                  [f'Has_{u.replace(" ","_")}' for u in usages_importants]].head(5))
```

```python
# 6. Features sur sources d'énergie :
# On crée des colonnes binaires indiquant si une source d'énergie est présente dans le bâtiment.
# Exemple :
#  Electricity(kBtu) = 0 → HasElectricity = 0
#  Electricity(kBtu) > 0 → HasElectricity = 1
# on fait ça car :
#  - Les modèles de machine learning ne doivent pas voir les valeurs exactes de consommation
#    (cela serait du "data leakage" si on prédit la consommation énergétique).
#  - Les colonnes binaires indiquent simplement la présence ou l'absence de chaque source.
#  - Cela permet au modèle de capturer l'effet de la présence d'une source sur la consommation ou d'autres cibles.
#  - Exemple pratique :
#       un bâtiment avec électricité et gaz naturel pourra consommer différemment d'un bâtiment qui n'a que de la vapeur.

batiments['HasElectricity'] = (batiments['Electricity(kBtu)'] > 0).astype(int)
batiments['HasNaturalGas'] = (batiments['NaturalGas(kBtu)'] > 0).astype(int)
batiments['HasSteam'] = (batiments['SteamUse(kBtu)'] > 0).astype(int)

print("Features 'HasElectricity', 'HasNaturalGas', 'HasSteam' ajoutées : indiquent la présence de chaque source d'énergie sans fuite de données")

display(batiments[['Electricity(kBtu)','NaturalGas(kBtu)','SteamUse(kBtu)',
                  'HasElectricity','HasNaturalGas','HasSteam']].head(3))
```

```python
# 7. Vérification finale des features :
# Après avoir créé toutes les nouvelles colonnes (BuildingAge, BuildingSize, NbPropertyUses,
# Has_<Usage>, HasElectricity, HasNaturalGas, HasSteam, etc.), il est important de vérifier
# que tout a été ajouté correctement et que les valeurs semblent cohérentes.
print("\nAperçu final des nouvelles features :")
display(batiments.head(3))
```

```python
# 8. Statistiques descriptives pour les features numériques :
# Après avoir créé toutes les nouvelles colonnes, il est utile de vérifier certaines statistiques
# de base pour comprendre la distribution et détecter d'éventuelles anomalies.
#
# Cela permet de : 
#  - Voir les valeurs minimales, maximales, moyennes et médianes.
#  - Détecter des erreurs évidentes (ex : âge négatif, taille de bâtiment absurde).
#  - Donné une idée générale de la répartition des données (variance, étendue).
#  - Un contrôle supplémentaire avant de lancer la modélisation.
print("\nStatistiques descriptives des features numériques :")
display(batiments[['BuildingAge','BuildingSize','NbPropertyUses','HasElectricity','HasNaturalGas','HasSteam']].describe())
```

```python
# ============================================================
# ÉTAPE 3 — PRÉPARATION DES FEATURES POUR LA MODÉLISATION
# ============================================================

# 1. Sélection des colonnes utiles pour l’entraînement
# On fait ça car :
# - Nous ne voulons garder que les colonnes vraiment utiles pour notre modèle.
# - On enlève les identifiants, les adresses, les textes, les infos inutiles.
# - On garde nos nouvelles features créées + colonnes numériques utiles.
# - L’objectif est d’éviter que le modèle soit perturbé par des données inutiles.

colonnes_a_garder = [
    'BuildingAge', 
    'BuildingSize',
    'NbPropertyUses',
    'HasElectricity',
    'HasNaturalGas', 
    'HasSteam',
    'Latitude', 'Longitude'
]

# On crée un nouveau DataFrame qui contient uniquement les colonnes sélectionnées
X = batiments[colonnes_a_garder]

print("Colonnes retenues pour le modèle :")
print(colonnes_a_garder)

display(X.head(3))  # Aperçu des données utilisées pour l'entraînement

```

```python
# 2. Nettoyage des outliers avec méthode des quantiles
# Pourquoi utiliser les quantiles ?
# - Certaines valeurs peuvent être extrêmement grandes (ex : bâtiments énormes ou l’âge est aberrant).
# - Ces valeurs extrêmes peuvent perturber le modèle.
# - La méthode des quantiles est plus flexible que l'IQR si on veut garder un maximum de données.
# - On va garder les valeurs entre le 1er et 99ème percentile pour chaque feature continue.

colonnes_numeriques = ['BuildingAge', 'BuildingSize']

# On crée une copie du DataFrame X pour éviter de modifier les données originales.
# X_clean sera la version nettoyée, tandis que X restera intact pour d'éventuelles vérifications.
X_clean = X.copy()

# Boucle sur chaque colonne numérique pour appliquer la suppression des valeurs extrêmes (outliers).
for col in colonnes_numeriques:
    # valeur sous laquelle se trouvent les 1% les plus faibles
    lower = X_clean[col].quantile(0.01)
     # valeur au-dessus de laquelle se trouvent les 1% les plus fortes
    upper = X_clean[col].quantile(0.99)
    # On mémorise la taille du dataset avant suppression, pour mesurer l’impact du nettoyage.
    old_len = len(X_clean)
    # On filtre le DataFrame pour ne conserver que les valeurs situées entre lower et upper.
    # Les lignes contenant des valeurs extrêmes pour cette colonne seront supprimées.
    X_clean = X_clean[(X_clean[col] >= lower) & (X_clean[col] <= upper)]
    # Calcul du pourcentage de lignes supprimées pour cette colonne utile pour vérifier qu'on ne
    # perd pas trop de données.
    suppression_ratio = (old_len - len(X_clean)) / old_len * 100
    # On affiche le résultat du nettoyage pour suivre combien de lignes ont été supprimées
    print(f"➡️ Nettoyage outliers pour {col} : {old_len} => {len(X_clean)} lignes "
          f"({suppression_ratio:.2f}% supprimées)")

print("\nAperçu après nettoyage des outliers :")
display(X_clean.head(3))
```

```python
# 3. Scaling des features continues avec RobustScaler
# Pourquoi RobustScaler ?
# - Il met toutes les colonnes sur une échelle comparable pour que le modèle comprenne mieux les différences entre elles.
# - Il est robuste : il n'est pas influencé par les valeurs très grandes ou très petites.
# - Il utilise la médiane et l’écart entre les 25% et 75% des valeurs pour calculer l'échelle.
# - On scale seulement les colonnes continues, pas les colonnes qui sont déjà 0/1.

from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

# Sélection des colonnes continues pour le scaling
colonnes_a_scaler = ['BuildingAge', 'BuildingSize', 'Latitude', 'Longitude']

# On applique le scaling
donnees_scalees = scaler.fit_transform(X_clean[colonnes_a_scaler])

# On reconstruit un DataFrame
X_scaled = X_clean
X_scaled[colonnes_a_scaler] = donnees_scalees

# X_scaled contient :
# - les colonnes continues normalisées (BuildingAge, BuildingSize, Latitude, Longitude)
# - les colonnes binaires inchangées (HasElectricity, HasNaturalGas, HasSteam)
print("\nDonnées après scaling avec RobustScaler (basé sur les quantiles) :")
display(X_scaled.head(3))
```

```python
# ------------------------------------------------------------
# 4. Vérification des distributions (statistiques descriptives)
# ------------------------------------------------------------
# Pourquoi ?
# - Avant d’entraîner un modèle, il est crucial de comprendre la distribution des features.
# - Cela permet de voir :
#     * si des colonnes sont déséquilibrées (ex : HasSteam = 0 presque partout)
#     * si certaines valeurs posent problème
#     * si une normalisation sera nécessaire plus tard

print("\n Statistiques descriptives après nettoyage :")
display(X_clean.describe())
```

```python

# ------------------------------------------------------------
# 5. Séparation entre X (features) et y (target)
# ------------------------------------------------------------
# Pourquoi ?
# - X = les colonnes utilisées pour prédire
# - y = la valeur que le modèle doit prédire
# - Ici, la target est : SiteEnergyUse(kBtu)

y = batiments.loc[X_clean.index, 'SiteEnergyUse(kBtu)']

print("\nSéparation X / y effectuée :")
print("X shape =", X_clean.shape)
print("y shape =", y.shape)
```

```python
#Étape 1 a 3 de l'exercice 

#Étape 1 : J’ai d’abord chargé et exploré le dataset pour comprendre les colonnes disponibles et identifier les valeurs manquantes. J’ai ensuite filtré les types de bâtiments les plus fréquents pour éviter le bruit.

#Étape 2 : j’ai fait du feature engineering : j’ai créé l’âge du bâtiment, renommé la surface, encodé les types de bâtiments en OneHotEncoder, construit des variables d’usages et d’énergie, et conservé les coordonnées géographiques.

#Étape 3 : Pour préparer la modélisation, j’ai sélectionné uniquement les colonnes pertinentes, nettoyé les valeurs extrêmes avec une méthode de quantiles 1%-99%, puis appliqué un scaling robuste avec RobustScaler.
# Et enfin, j’ai séparé X et y pour définir la target : la consommation énergétique SiteEnergyUse(kBtu). 
```

```python
# ============================================================
# ÉTAPE 4 — COMPARAISON DE DIFFÉRENTS MODÈLES SUPERVISÉS
# ============================================================

# Objectif de l'étape :
# - Tester plusieurs modèles pour prédire la consommation d'énergie (SiteEnergyUse(kBtu))
# - Comprendre les performances de chaque modèle à l'aide de métriques (MAE, R²)
# - Visualiser les résultats pour mieux interpréter les modèles
# - Décider quel modèle est le plus adapté pour ce problème

from sklearn.model_selection import train_test_split, cross_validate
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import make_scorer, r2_score, mean_absolute_error
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
# ------------------------------------------------------------
# 1. SÉPARATION TRAIN / TEST
# ------------------------------------------------------------
# Pourquoi cette étape est cruciale ?
# - Le modèle doit "apprendre" sur certaines données et être testé sur d'autres
#   pour vérifier s'il peut généraliser à de nouvelles données.
# 
# - Train (X_train, y_train) :
#     → Contient les données utilisées pour entraîner le modèle.
#     → Le modèle analyse les relations entre chaque colonne d'entrée (features)
#       et la colonne cible (target).
#     → Exemple : LinearRegression calcule ses coefficients pour minimiser l'erreur,
#       DecisionTree crée des règles pour séparer les valeurs, RandomForest combine plusieurs arbres.
# 
# - Test (X_test, y_test) :
#     → Contient des données que le modèle n'a jamais vues pendant l'apprentissage.
#     → Sert à mesurer la performance réelle du modèle (précision, erreurs, R², etc.).
#     → Permet de savoir si le modèle est capable de généraliser ou s'il est sur-appris (overfitting).
# 
# - random_state :
#     → Permet de toujours découper les données de la même manière en train et test.
#     → Chaque fois que vous exécutez le code avec le même random_state,
#       vous obtenez exactement les mêmes données dans X_train, X_test, y_train et y_test.
#     → Cela aide à comparer les résultats et à être sûr que le modèle est évalué
#       sur les mêmes données à chaque fois.

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,  # Features normalisées et préparées
    y,         # Target : SiteEnergyUse(kBtu)
    test_size=0.2,  # 20% des données pour tester, 80% pour entraîner
    random_state=0
)

print(" Séparation Train/Test effectuée")
print("   → X_train :", X_train.shape)
print("   → X_test  :", X_test.shape)
print("   → y_train :", y_train.shape)
print("   → y_test  :", y_test.shape)
```

```python
# ------------------------------------------------------------
# 2. DÉFINITION DES MÉTRIQUES
# ------------------------------------------------------------
# Pourquoi ces métriques MAE ET R2 ?
# - MAE ce dit (Mean Absolute Error) : moyenne des écarts absolus entre valeurs réelles et prédites
#   → donne une idée de l'erreur moyenne
# - R2 (coefficient de détermination) : proportion de la variance expliquée par le modèle
#   → R2 proche de 1 : le modèle explique bien les variations
#   → R2 proche de 0 : le modèle n’explique presque rien
scoring = {
    "MAE": make_scorer(mean_absolute_error),
    "R2": make_scorer(r2_score)
}
```

```python
# ------------------------------------------------------------
# 3. FONCTION POUR ÉVALUER UN MODÈLE
# ------------------------------------------------------------
# Pourquoi faire une fonction ?
# - Pour ne pas répéter le même code pour chaque modèle.
# - Pour centraliser l’évaluation, y compris la validation croisée.
# - La validation croisée (cross_validate) permet d’avoir une estimation plus fiable
#   de la performance du modèle en testant plusieurs sous-ensembles des données.
def evaluate_model(model, X_train, y_train):
    """
    Évalue un modèle sur les données d'entraînement via cross_validate
    et retourne les scores moyens MAE et R2.
    """
    print(f"\n Évaluation du modèle : {model.__class__.__name__}")
    print("   → Validation croisée (5 folds)...")

    # cross_validate : évalue le modèle sur 5 sous-ensembles pour obtenir une moyenne plus fiable
    cv_results = cross_validate(
        model,
        X_train,
        y_train,
        cv=5,  # 5 folds : diviser le train en 5 parties, chacune sert de test à tour de rôle
        scoring=scoring,
        return_train_score=False  # on s'intéresse seulement aux performances sur les "folds" de test
    )

    # Calcul des moyennes sur les 5 folds
    mae_mean = cv_results["test_MAE"].mean()
    r2_mean = cv_results["test_R2"].mean()

    print("   → MAE moyen :", round(mae_mean, 3))
    print("   → R2 moyen  :", round(r2_mean, 3))

    # Calcul des scores moyens :
    # - MAE : moyenne des erreurs absolues → combien le modèle se trompe en moyenne
    # - R²  : proportion de la variance expliquée par le modèle
    # La fonction renvoie ensuite un dictionnaire avec :
    #   * "Modèle" : nom du modèle
    #   * "MAE" : valeur moyenne MAE
    #   * "R2" : valeur moyenne R²    
    return {
        "Modèle": model.__class__.__name__,
        "MAE": mae_mean,
        "R2": r2_mean
    }
```

```python
# ------------------------------------------------------------
# 4. LISTE DES MODÈLES À TESTER
# ------------------------------------------------------------
# Pourquoi j'utilise ces modèles ?
# - LinearRegression : modèle simple et interprétable → point de référence
# - DecisionTreeRegressor : modèle non-linéaire simple → visualisation des décisions
# - RandomForestRegressor : ensemble d’arbres → souvent plus précis et moins sensible au surapprentissage
models = [
    LinearRegression(),
    DecisionTreeRegressor(random_state=0),
    RandomForestRegressor(random_state=0)
]

print("\n Modèles testés :")
for m in models:
    print("   -", m.__class__.__name__)
```

```python
# ------------------------------------------------------------
# 5. ÉVALUATION ET VISUALISATION DES MODÈLES
# ------------------------------------------------------------
results = [] 

    
    # Boucle pour évaluer et visualiser tous les modèles
for model in models:

    # a) ÉVALUATION AVEC CROSS_VALIDATE 
    # Pourquoi utiliser cross_validate ?
    # - Principalement pour avoir une estimation plus fiable de la performance du modèle.
    # - On divise X_train et y_train en 5 sous-ensembles on appel ça des (folds)
    # - Chaque fold sert de test à tour de rôle, les 4 autres servent pour l'entraînement
    # - On calcule MAE et R² pour chaque fold et on fait la moyenne
    # - Cela permet de savoir si le modèle généralise bien sans utiliser X_test
    scores = evaluate_model(model, X_train, y_train)

    # On stocke les résultats pour ce modèle dans une liste
    results.append(scores)

    # b) ENTRAÎNEMENT COMPLET SUR X_train
    # Pourquoi fit ici après cross_validate ?
    # - cross_validate ne modifie pas le modèle, il l’entraîne "temporairement" sur chaque fold
    # - Pour faire des prédictions ou visualisations après, il faut entraîner le modèle sur tout X_train
    model.fit(X_train, y_train)
    # Ici avec fit(), le modèle "apprend" :
    # - Il analyse les relations entre toutes les colonnes de X_train (features)
    # - Et la colonne y_train (target)
    # - Chaque modèle a sa façon d'apprendre :
    #     * LinearRegression : calcule les coefficients pour minimiser l'erreur
    #     * DecisionTree : crée des règles (conditions sur les features) pour séparer les valeurs
    #     * RandomForest : construit plusieurs arbres et combine les décisions


    # c) Visualisation selon le type de modèle
    # LinearRegression : graphique qui compare les vraies valeurs de TEST
    # aux valeurs prédites par le modèle.
    # Si les points sont proches de la ligne rouge, cela signifie que
    # les prédictions du modèle sont bonnes.   
    if isinstance(model, LinearRegression):
        y_pred = model.predict(X_test)  # on prédit les valeurs sur TEST
        plt.figure(figsize=(6,5))
        plt.scatter(y_test, y_pred, alpha=0.5)
        plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--')  # ligne parfaite
        plt.xlabel("Valeurs réelles")
        plt.ylabel("Valeurs prédites")
        plt.title("LinearRegression : Prédiction vs Réel")
        plt.text(
            x=y_test.min(),
            y=y_test.max()*0.9,
            s="Points proches de la ligne rouge → bonnes prédictions",
            color='blue'
        )
        plt.show()

        # DecisionTreeRegressor : visualisation de l’arbre de décision appris par le modèle.
        # Ici, on n’utilise pas predict(), car on ne veut pas voir les prédictions,
        # mais comprendre COMMENT le modèle décide.
        # Chaque bloc représente une question posée sur une feature
        # (ex : “BuildingSize < 5000 ?”).
        # En suivant les branches, on voit les règles que l’arbre applique
        # pour arriver à une prédiction finale.
        # Ce graphique permet de comprendre quelles features influencent le plus
        # et comment le modèle sépare les données.
    elif isinstance(model, DecisionTreeRegressor):
        plt.figure(figsize=(14,8))
        plot_tree(
            model,
            max_depth=2,
            feature_names=X_train.columns,
            filled=True,
            fontsize=10
        )
        plt.title("DecisionTreeRegressor : Structure de l'arbre")
        plt.show()
        
        # RandomForestRegressor : visualisation de l’importance des features.
        # Ici aussi on n’utilise pas predict(), car l’objectif est de comprendre quelles colonnes
        # (features) influencent le plus les prédictions du modèle.
        # Chaque barre du graphique représente une feature.
        # La hauteur de la barre indique l’importance de cette feature :
        # plus la barre est haute, plus la colonne impacte la prédiction de la consommation d’énergie.
        # Cela permet de savoir quelles variables le modèle considère comme les plus pertinentes
        # pour expliquer SiteEnergyUse(kBtu).
    elif isinstance(model, RandomForestRegressor):
        importances = model.feature_importances_
        indices = np.argsort(importances)[::-1]  # trier du plus important au moins important
        plt.figure(figsize=(8,5))
        plt.bar(range(X_train.shape[1]), importances[indices])
        plt.xticks(range(X_train.shape[1]), X_train.columns[indices], rotation=45)
        plt.title("RandomForestRegressor : Importance des features")
        plt.text(
            0,
            max(importances)*0.8,
            "Hauteur = influence de chaque feature sur la prédiction",
            color='blue'
        )
        plt.show()
```

```python
# ------------------------------------------------------------
# 6. RÉSULTATS FINALS
# ------------------------------------------------------------
results_df = pd.DataFrame(results)
print("\n Résultats finaux (moyenne sur validation croisée) :")
display(results_df)
```

```python
# ------------------------------------------------------------
# 7. INTERPRÉTATION
# ------------------------------------------------------------
# - MAE bas → modèle précis
# - R² haut → modèle explique bien la variance
# - LinearRegression : points proches de la diagonale → fiable
# - DecisionTree : comprendre les décisions prises par l'arbre
# - RandomForest : voir quelles features influencent le modèle
# - Étape utile pour comparer modèles et choisir le plus adapté
```

```python
# ============================================================
# ÉTAPE 5 : OPTIMISATION ET INTERPRÉTATION DU MODÈLE
# ============================================================

# Dans cette étape, l'objectif est d'améliorer les performances du modèle
# en testant plusieurs combinaisons d'hyperparamètres grâce à GridSearchCV.
# Ensuite, une fois le meilleur modèle trouvé, nous analyserons l'importance
# des variables (features importance) pour comprendre ce qui influence le plus
# la consommation énergétique.

from sklearn.model_selection import GridSearchCV

# ------------------------------------------------------------
# 5.1 - DÉFINITION D'UNE PETITE GRILLE DE TEST (pour vérifier le code)
# ------------------------------------------------------------
# Pourquoi une petite grille ?
# - Avant de tester des centaines de combinaisons, il faut s'assurer 
#   que GridSearchCV fonctionne bien et que votre code ne plante pas.
# - Cette petite grille permet d'exécuter seulement ~8 combinaisons.

param_grid_small = {
    "n_estimators": [50, 100],     # nombre d'arbres dans la forêt
    "max_depth": [5, 10],          # profondeur maximale des arbres
    "min_samples_split": [2, 5],   # minimum d'échantillons pour diviser un nœud
}

print("🔍 Test de la petite grille pour vérifier que tout fonctionne...")

grid_small = GridSearchCV(
    estimator=RandomForestRegressor(random_state=42),
    param_grid=param_grid_small,
    cv=5,                     # Validation croisée en 5 folds
    scoring="neg_mean_absolute_error",
    n_jobs=-1                 # utilise tous les CPU disponibles
)

grid_small.fit(X_train, y_train)

print("Petite grille OK ✔️")
print("Meilleurs paramètres (petite grille) :", grid_small.best_params_)


# ------------------------------------------------------------
# 5.2 - DÉFINITION DE LA GRANDE GRILLE À TESTER (optimisation réelle)
# ------------------------------------------------------------
# Maintenant que tout fonctionne, on peut lancer une grille plus large.
# Attention : trop de combinaisons = très long, il faut rester raisonnable.
# L'objectif recommandé : entre 100 et 500 combinaisons.

param_grid = {
    "n_estimators": [200, 400, 600],
    "max_depth": [10, 20, 30],
    "min_samples_split": [2, 5, 10],
    "max_features": ["sqrt", "log2"]
}
# Total = 3 × 3 × 3 × 2 = 54 combinaisons → raisonnable et rapide


print("🚀 Lancement de la vraie GridSearchCV... Cela peut prendre un moment.")

grid = GridSearchCV(
    estimator=RandomForestRegressor(random_state=42),
    param_grid=param_grid,
    cv=5,
    scoring="neg_mean_absolute_error",
    n_jobs=-1,
    verbose=1      # montre la progression
)

grid.fit(X_train, y_train)

print("✔️ GridSearch terminée")
print("Meilleurs paramètres trouvés :", grid.best_params_)


# ------------------------------------------------------------
# 5.3 - RÉCUPÉRATION DU MEILLEUR MODÈLE
# ------------------------------------------------------------
best_model = grid.best_estimator_
print("Modèle optimisé chargé ✔️")

# Prédictions sur le test
y_pred_best = best_model.predict(X_test)

# Calcul des performances
mae_best = mean_absolute_error(y_test, y_pred_best)
r2_best = r2_score(y_test, y_pred_best)

print("\n📊 Performance du modèle optimisé :")
print("MAE :", mae_best)
print("R²  :", r2_best)


# ------------------------------------------------------------
# 5.4 - FEATURE IMPORTANCE (interprétation du modèle)
# ------------------------------------------------------------
# Maintenant que le meilleur modèle est trouvé, nous pouvons analyser
# quelles colonnes ont le plus d'influence sur la prédiction.

importances = best_model.feature_importances_
indices = np.argsort(importances)[::-1]  # tri décroissant

plt.figure(figsize=(10,6))
plt.bar(range(len(importances)), importances[indices])
plt.xticks(range(len(importances)), X_train.columns[indices], rotation=90)
plt.title("RandomForest Optimisé : Importance des variables")
plt.tight_layout()
plt.show()

print("\nLes features les plus influentes ont été affichées dans le graphique.")
```
